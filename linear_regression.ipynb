# Jupyter Notebook в VS Code 

## Ячейки

### Как создавать ячейки  
В Jupyter Notebook (VS Code) ячейку можно создать несколькими способами:

#### Кнопками  
Для этого нужно над или под ячейкой нажать кнопку `+ Code` для появления ячейки с кодом или `+ Markdown` для появления ячейки разметки).  

#### Горячими клавишами
Работают в **command mode** — сначала нужно нажать `Esc`, чтобы выйти из режима
редактирования текста:
- `A` — добавить ячейку **выше** текущей (Above)
- `B` — добавить ячейку **ниже** текущей (Below)
- `Alt + Enter` — выполнить текущую и создать новую снизу
- `Shift + Enter` — выполнить текущую ячейку и перейти к следующей

#### Правым кликом
Правой кнопка мыши -> **Insert Cell** -> выбрать нужную вставку:  
**Insert Code Cell Above** — ячейка кода выше  
**Insert Code Cell Below** — ячейка кода ниже  
**Insert Markdown Cell Above** — ячейка разметки выше  
**Insert Markdown Cell Below** — ячейка разметки ниже 

#### Изменение типа ячейки
Тип ячейки переключается в командном режиме:
- `Y` — Code
- `M` — Markdown
- `R` — Raw

Либо через выпадающий список в правом нижнем углу ячейки.

### Виды ячеек  
- **Ячейки кода (Code cells)** — нужны для написания и выполнения исполняемого кода. При запуске код выполняется ядром, результат (число, таблица, график) отображается прямо под ячейкой.
- **Ячейки разметки (Markdown cells)** —  нужны для оформления документа. В них можно писать обычный текст, делать заголовки, списки, таблицы, вставлять ссылки и математические формулы через синтаксис LaTeX.
- **Сырой текст (Raw NBConvert)** — ячейка, содержимое которой выводится на экран без какой-либо обработки со стороны Jupyter или языка программирования (используется редко, в основном для генерации документации).


### Способы запуска ячеек
- `Shift + Enter` — выполнить текущую ячейку и перейти к следующей.
- `Ctrl + Enter` — выполнить текущую ячейку и остаться в ней.
- `Alt + Enter` — выполнить текущую ячейку и создать новую ячейку под ней.
- Нажать кнопку ▶ слева от ячейки для запуска текущей ячейки.
- В меню над ячейкой нажать кнопку ▶ со стрелкой вверх для запуска всех ячеек сверху(не включая текущую), ▶ со стрелкой вниз для запуска всех ячеек ниже(включая текущую) 
- В верхнй панели над рабочей областью нажать **Run All**. Выполнится весь Notebook сверху вниз.



### Влияние ячеек друг на друга
Ячейки выполняются последовательно, и все переменные, объявленные в предыдущих ячейках, доступны в последующих. Если перезапустить ядро (Restart Kernel), все переменные исчезнут. Если выполнять ячейки в произвольном порядке, то тогда могут возникнуть ошибки или быть получены неверные результаты.

## Markdown

### Что такое Markdown

Markdown — это лёгкий язык разметки, который позволяет простым текстом задавать форматирование (заголовки, списки, ссылки, изображения, выделение, код, формулы).

### Как создавать заголовки и просматривать их
Создаются символом `#` в начале строки. Чем больше `#`, тем ниже уровень:

**# H1** — заголовок первого уровня
**## H2** — второго
**### H3** — третьего и т.д.

Для просмотра заголовков в VS Code используется панель **Outline**. Она находится в левой панели `Explorer` в выпадаюшем меню `Outline`

### Жирный и курсив

- `**жирный**` → **жирный**
- `*курсив*` → *курсив*
- `***жирный курсив***` → ***жирный курсив***
- `~~зачёркнутый~~` → ~~зачёркнутый~~

### Код с подсветкой синтаксиса

Блок кода задаётся тройными обратными кавычками  с указанием языка:

````markdown
```python
def square(x):
    return x ** 2

print(square(5))
```
````
Как будет выглядеть:
```python
def square(x):
    return x ** 2

print(square(5))
```

Инлайн-код задаётся одинарными обратными кавычками `` `print("hello")` `` -> `print("hello")`

### Формулы в записи LaTeX

Для записи в строке используется знаки доллара `$y = b_0 + b_1 x_1 + b_2 x_2$` → $y = w_0 + w_1 x_1 + w_2 x_2$

Для записи отдельным блоком используются двойные доллары в начале и в конце:
```latex
$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$
```
Что получиться:  
$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

Ещё примеры:

- дробь: `` `$\frac{a}{b}$` `` → $\frac{a}{b}$
- корень: `` `$\sqrt{x}$` `` → $\sqrt{x}$
- сумма: `` `$\sum_{i=1}^{n} x_i$` `` → $\sum_{i=1}^{n} x_i$
- интеграл: `` `$\int_{a}^{b} f(x)\,dx$` `` → $\int_{a}^{b} f(x)\,dx$
- греческие буквы: `` `$\alpha, \beta, \gamma, \sigma, \mu$` `` → $\alpha, \beta, \gamma, \sigma, \mu$
- индекс и степень: `` `$x_i^2$` `` → $x_i^2$

### 7-point summary

`describe()` класса `pandas.DataFrame` возвращает по каждому числовому признаку 8 характеристик:

- **count** — количество непустых значений;
- **mean** — среднее арифметическое;
- **std** — стандартное отклонение (разброс относительно среднего);
- **min** — минимум;
- **25%** — первый квартиль;
- **50%** — медиана (середина распределения);
- **75%** — третий квартиль;
- **max** — максимум.

#### Выводы
- **Пропусков нет**: `count` = 3921 у всех признаков.

- **`yr_built`** — узкий диапазон (1900–2015), выбросов нет.

`````python
# Импорт библиотек
import seaborn
import matplotlib.pyplot as plt

# Считаем, сколько нужно строк и столбцов для сетки графиков
ncols = 3                          # по 3 графика в строке
n = len(df0.columns)               # сколько всего признаков (у тебя 16)
nrows = -(-n // ncols)             # округление вверх: ceil(n / 3) = 6 строк

# Создаём полотно: nrows строк × ncols колонок, размер 18 × 4*nrows дюймов
# fig — полотно, axes — двумерный массив осей (по одной на каждый график)
fig, axes = plt.subplots(nrows, ncols, figsize=(18, 4 * nrows))

# Превращаем двумерный массив осей в одномерный — удобнее обращаться по индексу i
axes = axes.flatten()

# Проходим по всем столбцам df0 (i — индекс, col — имя столбца)
for i, col in enumerate(df0.columns):
    # Строим boxplot для текущего признака в i-й ячейке
    seaborn.boxplot(y=df0[col], ax=axes[i])
    # Заголовок ячейки = имя признака
    axes[i].set_title(col)

# Скрываем лишние пустые ячейки (16 столбцов → 6×3 = 18 ячеек, 2 лишние)
for j in range(i + 1, len(axes)):
    axes[j].axis("off")

# Убираем наложение подписей между графиками
plt.tight_layout()
# Показываем полотно
plt.show()
`````

`````python
# Импорт библиотек
import seaborn
import matplotlib.pyplot as plt

# Считаем, сколько нужно строк и столбцов для сетки графиков
ncols = 3                          # по 3 графика в строке
n = len(df0.columns)               # сколько всего признаков (у тебя 16)
nrows = -(-n // ncols)             # округление вверх: ceil(n / 3) = 6 строк

# Создаём полотно: nrows строк × ncols колонок
fig, axes = plt.subplots(nrows, ncols, figsize=(18, 4 * nrows))

# Превращаем двумерный массив осей в одномерный
axes = axes.flatten()

# Проходим по всем столбцам df0
for i, col in enumerate(df0.columns):
    # Строим violinplot для текущего признака
    seaborn.violinplot(y=df0[col], ax=axes[i])
    axes[i].set_title(col)

# Скрываем лишние пустые ячейки
for j in range(i + 1, len(axes)):
    axes[j].axis("off")

plt.tight_layout()
plt.show()
`````

## Boxplot и violinplot

**Boxplot (диаграмма размаха)** — график, который показывает распределение значений.
На нём есть прямоугольник и линии-усы сверху и снизу:

- **горизонтальная линия внутри прямоугольника — медиана** (середина данных)
- **прямоугольник** — от 25 % до 75 % значений
- **вертикальные линии-усы сверху и снизу от прямоугольника** — границы нормальных значений
- **точки за усами — выбросы** (нетипично большие или маленькие значения)

**Violinplot (скрипичная диаграмма)** — то же самое, но ещё показывает
плотность. Если фигура широкая, то там много значений, если узкая — мало.

Линейная регрессия чувствительна к выбросам. Они могут сильно исказить
коэффициенты модели и ухудшить предсказания для всех остальных. Такие
значения стоит проверить или ограничить.

`````python
# bins="auto" — библиотека сама подбирает рекомендованное число столбцов
# kde=True — добавляет сглаженную кривую плотности
seaborn.histplot(df0["price_ln"], bins="auto", kde=True)
plt.title("Гистограмма price_ln (bins='auto')")
plt.show()
`````

`````python
# Много столбцов (300)
seaborn.histplot(df0["price_ln"], bins=300, kde=True)
plt.title("Гистограмма price_ln (bins=300)")
plt.show()
`````

### Гистограмма целевого признака `price_ln`

### Как устроена гистограмма

Гистограмма берёт диапазон значений `price_ln` (примерно от 11.5 до 15.5) и разбивает его на **интервалы** (`bins`). По вертикали — **частота**: сколько домов попало в каждый интервал. Чем больше `bins`, тем мельче интервалы и детальнее видна структура распределения.

### Что видно при большом числе столбцов (`bins=300`)

- Распределение **близко к симметричному**, форма напоминает колокол — это результат **логарифмирования цены**. Исходная цена без логарифма была бы сильно скошена вправо, а `price_ln` почти нормален.
- Центр распределения — около **13.0**, что совпадает с медианой из `describe()` (13.02).
- Лёгкая **правосторонняя асимметрия** всё же сохраняется — «хвост» справа немного длиннее, чем слева.
- Мелкие **всплески и провалы** на столбцах — результат округлений в исходных данных.

### Вывод

Распределение `price_ln` **почти нормальное**. Это хорошо для линейной регрессии: МНК предполагает нормальные остатки, и с таким целевым признаком модель работает лучше, чем с исходной ценой.

### Матрица корреляции

Корреляция показывает линейную связь двух признаков.
Значения от $-1$ до $+1$:
- $+1$ — сильная прямая связь;
- $-1$ — сильная обратная;
- $0$ — линейной связи нет.

### Какие признаки влияют на `price_ln` сильнее всего

По корреляции с целевым:

- **grade — 0.69**. Самый сильный признак: чем выше класс отделки дома, тем дороже.
- **sqft_living — 0.68**. Сильная прямая связь: больше жилая площадь — выше цена.
- **sqft_living15 — 0.63**. Средняя площадь соседних домов тоже влияет на цену.
- **sqft_above — 0.59**. Площадь над землёй.
- **bathrooms — 0.54**. Число ванных.
- **lat — 0.44**. Широта (географическое положение).
- **view — 0.34**, **bedrooms — 0.33**, **sqft_basement — 0.29**, **floors — 0.29** — умеренная связь.
- **waterfront — 0.21**, **sqft_lot — 0.11** — слабая связь.
- **yr_built — 0.08**, **long — 0.05**, **condition — 0.05** — почти нет связи с ценой.

Значит, основной вклад в регрессию дадут **grade**, **sqft_living**,
**sqft_living15**, **sqft_above**, **bathrooms**.

### Есть ли зависимость между независимыми признаками

Да, есть, и довольно сильная:

- **sqft_living ↔ sqft_above — 0.87**. Самая сильная связь: почти вся жилая
  площадь находится над землёй.
- **sqft_living ↔ sqft_living15 — 0.76**. Дом похож по площади на соседние.
- **sqft_living ↔ bathrooms — 0.75**. Больше площадь — больше ванных.
- **sqft_above ↔ sqft_living15 — 0.74**. Связано с площадью.
- **grade ↔ sqft_living — 0.76**, **grade ↔ sqft_above — 0.75**,
  **grade ↔ sqft_living15 — 0.71**. Класс отделки растёт вместе с площадью.
- **bathrooms ↔ sqft_above — 0.69**, **bathrooms ↔ grade — 0.66**.
- **bedrooms ↔ sqft_living — 0.58**, **bedrooms ↔ bathrooms — 0.52**.

Все эти пары — **мультиколлинеарность**.

### Как это влияет на линейную регрессию

**Мультиколлинеарность** — плохо для линейной регрессии:

- коэффициенты модели становятся **неустойчивыми** — небольшие изменения
  в данных сильно их меняют;
- модель не понимает, какому из связанных признаков отдать вклад;
- коэффициенты теряют смысл, их нельзя нормально интерпретировать.

**Что делать:**
- убрать один из пары связанных признаков (например, `sqft_above` или
  `sqft_living15`, оставив `sqft_living`);
- использовать **Ridge** или **Lasso**;
- использовать **PCA**.

### Pairplot

`pairplot` строит **матрицу попарных диаграмм рассеяния**. На пересечении
строки и столбца — scatter для двух признаков, а по диагонали — распределение
каждого признака отдельно.

Что показывает:
- **линейные зависимости** между парами признаков;
- **кластеры** — группы похожих объектов;
- **выбросы** — отдельные точки в стороне;
- **форму распределения** каждого признака (по диагонали).

### Что видно на pairplot

- **`price_ln`** (целевой) распределён **почти нормально** — колокол
  без резкой асимметрии. Это результат логарифмирования цены:
  исходная цена была бы сильно скошена вправо.
- Несколько признаков — **дискретные**: `bedrooms`, `bathrooms`, `floors`,
  `waterfront`, `view`, `condition`, `grade`. На scatter они образуют
  вертикальные и горизонтальные «полосы», потому что принимают
  ограниченный набор значений.
- **Сильная линейная связь** между парами: `sqft_living` ↔ `sqft_above`,
  `sqft_living` ↔ `sqft_living15`, `sqft_living` ↔ `bathrooms`,
  `grade` ↔ `sqft_living`.
- **`lat` и `long`** связаны слабее, но scatter-облака имеют структуру:
  видно, что дома группируются по географическим районам.
- **`yr_built`** показывает вертикальные полосы — есть периоды массовой
  застройки.

### Для какого распределения `price_ln` предсказания точнее

Линейная регрессия (МНК) предполагает, что остатки нормальны, а связь
$y$ с признаками линейна. Значит, точнее всего предсказания при
**нормальном распределении целевого признака**.

У тебя `price_ln` — **как раз почти нормальное распределение** (это видно
и на pairplot, и на гистограмме). Это хорошо: логарифмирование цены уже
сделано заранее, поэтому модель будет работать точнее, чем на исходной
цене без логарифма.
